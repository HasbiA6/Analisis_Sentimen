```python
from google_play_scraper import reviews, Sort

scrapreview = reviews(
    'com.shopee.id',          # ID aplikasi
    lang='id',             # Bahasa ulasan (default: 'en')
    country='id',          # Negara (default: 'us')
    sort=Sort.MOST_RELEVANT, # Urutan ulasan (default: Sort.MOST_RELEVANT)
    count=10000           # Jumlah maksimum ulasan yang ingin diambil
)
```

```python
import csv

with open('ulasan_shopee.csv', mode='w', encoding='utf8', newline='') as file:
    writer = csv.writer(file)
    # Write the header row with all the keys from the first review
    if scrapreview and scrapreview[0]:
        header = scrapreview[0][0].keys() # Access the first element of the tuple, then the first dictionary
        writer.writerow(header)
        # Write the values for each review
        for review in scrapreview[0]: # Iterate through the list of reviews (first element of the tuple)
            writer.writerow(review.values())
```
